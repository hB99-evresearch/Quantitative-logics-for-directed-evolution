```python
def matrix_to_3d(wdec, matrix):

    if wdec==1:
        pio.renderers.default = 'browser' 
    if wdec==2:
        pio.renderers.default = 'notebook'
    depth, rows, cols = matrix.shape
    cmatrix=copy.deepcopy(matrix)
    stacked_matrix=np.array([])
    stacked_points=np.array([])
    for x in range(depth):
        temp_x_matrix=np.array(matrix_emptier(cmatrix[x,:,:]))
        temp_x_points=point_orderer(temp_x_matrix)
        stacked_matrix, stacked_points= three_dim_maker(stacked_matrix, stacked_points, temp_x_matrix, temp_x_points, x)

    three_dim_drawer(stacked_matrix, stacked_points)

def matrix_emptier(matrix):

    rows, cols= matrix.shape
    for i in range(rows):
        ones_indices = np.where(matrix[i] == 1)[0]
        for j in range(cols):
            if len(ones_indices) >= 2:
    # Keep only the first and last
                first = ones_indices[0]
                last = ones_indices[-1]

    # Set all 1s to 0 first
                matrix[i][matrix[i][j] == 1] = 0

    # Set only first and last 1s back to 1
                matrix[i][first] = 1
                matrix[i][last] = 1
    
    return matrix

def point_orderer(matrix):
    points = np.argwhere(matrix == 1)
    len_points=len(points)
    copy_points=copy.deepcopy(points)

    temp_points_array=[]

    for i in range(int(len(points)/2)):
        temp_points_array.append(points[2*i+1])
    temp_points_array=np.array(temp_points_array)
    copy_points=np.array(copy_points)
    temp_points_array=temp_points_array.tolist()
    copy_points=copy_points.tolist()


    for row in copy_points:
        if row in temp_points_array:
            copy_points.remove(row)

    first_half_points = copy_points
    second_half_points = temp_points_array

    for i in range(len(second_half_points)):
        first_half_points.append(second_half_points[int(len(points)/2)-(i+1)])

    new_points= np.array(first_half_points)
    return new_points

def three_dim_maker(stacked_matrix, stacked_points, matrix, points, x): #stack in x direction
    matrix=matrix.tolist()
    stacked_matrix=stacked_matrix.tolist()
    stacked_matrix.append(matrix)

    new_points=[]
    points=points.tolist()
    for el in points:
        new_points.append([x]+el)
        
    
    stacked_points=stacked_points.tolist()
    stacked_points+=new_points
    return np.array(stacked_matrix), np.array(stacked_points)





def three_dim_drawer(matrix, points):
    import plotly.graph_objects as go
    import numpy as np
    from itertools import combinations
    import copy
    from scipy.spatial import ConvexHull

    depth, rows, cols = matrix.shape
    point_x_dict = {}
    for x_val in range(depth):
        k = 0
        for point_el in points:
            if point_el[0] == x_val:
                k += 1
        if x_val == 0:
            point_x_dict[x_val] = k
        else:
            point_x_dict[x_val] = k + point_x_dict[x_val - 1]

    points = np.array(points)
    len_points = len(points)
    x, y, z = points[:, 0], points[:, 1], points[:, 2]

    # --- Original BLUE perimeter lines (unchanged) ---
    first_2d_array = []
    for el1 in range(depth):
        if el1 == 0:
            for elem1 in range(0, point_x_dict[0] - 1):
                first_2d_array.append((elem1, elem1 + 1))
                first_2d_array.append((0, point_x_dict[0] - 1))
        else:
            for elem1 in range(point_x_dict[el1 - 1], point_x_dict[el1] - 1):
                first_2d_array.append((elem1, elem1 + 1))
                first_2d_array.append((point_x_dict[el1 - 1], point_x_dict[el1] - 1))

    edge_traces = []
    for start, end in first_2d_array:
        edge_traces.append(go.Scatter3d(
            x=[x[start], x[end]],
            y=[y[start], y[end]],
            z=[z[start], z[end]],
            mode='lines',
            line=dict(color='blue', width=3),
            showlegend=False
        ))

    # --- NEW: GREEN perimeter lines for each y-plane (using convex hull) ---
    green_traces = []
    unique_y = np.unique(y)
    for y_val in unique_y:
        mask = y == y_val
        slice_points = np.column_stack((x[mask], y[mask], z[mask]))

        if len(slice_points) >= 3:
            # Project to (x,z)
            xz_points = slice_points[:, [0, 2]]

            try:
                hull = ConvexHull(xz_points)
                hull_indices = hull.vertices
                hull_loop = np.append(hull_indices, hull_indices[0])  # close loop

                for i in range(len(hull_loop) - 1):
                    i1, i2 = hull_loop[i], hull_loop[i + 1]
                    green_traces.append(go.Scatter3d(
                        x=[slice_points[i1, 0], slice_points[i2, 0]],
                        y=[slice_points[i1, 1], slice_points[i2, 1]],
                        z=[slice_points[i1, 2], slice_points[i2, 2]],
                        mode='lines',
                        line=dict(color='blue', width=3),
                        showlegend=False
                    ))
            except:
                # Not enough distinct points for a hull
                pass

    # --- Original vertex markers ---
    vertex_trace = go.Scatter3d(
        x=x, y=y, z=z,
        mode='markers+text',
        marker=dict(size=2, color='blue'),
        text=[f"P{i}" for i in range(len(x))],
        textposition="top center",
        name='Vertices'
    )

    # --- Combine everything ---
    fig = go.Figure(data=[vertex_trace] + edge_traces + green_traces)
    fig.update_layout(
        title="3D Shape with Blue (original) and Green (true y-plane perimeters)",
        scene=dict(
            xaxis=dict(title='X', range=[0, 15]),
            yaxis=dict(title='Y', range=[0, 15]),
            zaxis=dict(title='Z', range=[0, 15]),
            aspectmode='manual',
            aspectratio=dict(
                x=depth,
                y=rows *1.3,
                z=cols * 1.4    # broader z axis
            ),
    
            camera=dict(      # <-- MUST be inside scene
                eye=dict(x=25, y=25, z=25)
            )
        ),



        width=800,
        height=800
    )
    fig.show()




def tuple_to_np_array(tuple_algebra):
    new_algebra_of_interest = []
    for inner in tuple_algebra:
        inner_list = []
        for x in inner:
            inner_list.append(list(x))
        new_algebra_of_interest.append(inner_list)

    new_algebra_of_interest=np.array(new_algebra_of_interest)

    return new_algebra_of_interest


def tuple_to_list(obj):
    """Recursively convert all tuples to lists."""
    if isinstance(obj, tuple):
        return [tuple_to_list(x) for x in obj]
    elif isinstance(obj, list):
        return [tuple_to_list(x) for x in obj]
    else:
        return obj
```

```python
def potential(summed_function, algebra_dict, function,select_dict,s):
  temp_fit={}
  temp_pot={}
  temp_rate={}

  temp_ind=list(function.keys())
  temp_ind.sort()
  
  for i in temp_ind:
      temp_fit[i]=1+select_dict[i]

  
  weight=0
  
  for i in temp_ind:
      y=list(function[i])
      weight+=(float(temp_fit[i])*(float(y[0])))

  for i in temp_ind:
      temp_pot[i]=temp_fit[i]/weight

  for i in temp_ind:
      pot=float(temp_pot[i])
      y=list(function[i])

      if pot==1:
          pot_bar=1
      else:
          x=y[0]
          a=(1-pot*x)
          b=(1-x)
          pot_bar=a/b

      A=np.array([[pot,0],[0,pot_bar]])

      y_np=np.array(y)
      y_np=A@y_np
      z=y_np.tolist()
      function[i]=z

 
  return function
```

```python
def s_value_zeroer(function, select_dict,N,s):
    # Ensure consistent key order and numeric data
    established_function_key_array=[]
    for key in list(function.keys()):
        if function[key][0]>(1/(N*s)):
            established_function_key_array.append(key)
    
    crit_values = [float(select_dict[k]) for k in established_function_key_array]
    min_val = min(crit_values)

    
    key_array=list(function.keys())
    values = [float(select_dict[k]) for k in key_array]

    
    # Subtract the min safely and return mapping
    temporary_s_value_dictionary = {k: (v - min_val) for k, v in zip(key_array, values)}
    return temporary_s_value_dictionary
```

```python
def drift(function, N, N_ori, dynamic_mode): #drift D operator information could be explicitly expressed, but to reduce the computation complexity, only the result after D was expressed here.
  h={}
  ch_function=dict(sorted(function.items()))
  ftns=ch_function.values()
  temp_ind=list(function.keys())
  temp_ind.sort()
  length=len(temp_ind)

  for i in temp_ind:
    h[i]=math.floor(float(ch_function[i][0])*N*2)

  allocation_bound=list(h.values())
  
  seed = random.randint(0,2**32-1)
  gen = np.random.Generator(np.random.PCG64(seed))
    
  if dynamic_mode=='on':
    sigmoid_proportion=(1+(1/(math.exp(-1*math.tan(math.pi*(N/N_ori)/2))+1)))/2
    if N/N_ori<0.9:
      randomness=random.uniform(1,1+0.2*math.exp(-1*N/N_ori))
    else:
      randomness=random.uniform(1-0.2*math.exp(-1*math.tan(math.pi*(N/N_ori)/2)),1+0.2*math.exp(-1*math.tan(math.pi*(N/N_ori)/2)))
    
    new_allocation_bound=[]
    for i in allocation_bound:
      new_allocation_bound.append(math.floor(i*randomness*sigmoid_proportion))
    allocation_bound=np.array(new_allocation_bound)
    ordered_values=gen.multivariate_hypergeometric(allocation_bound, N)
    ordered_values=list(ordered_values)
  
  else:
    ordered_values=gen.multivariate_hypergeometric(allocation_bound, N)
    ordered_values=list(ordered_values)

  for i in range(length):
    ordered_values[i]=ordered_values[i]/N

  Y=[]
  for i in range(length):
    Y.append(i)

  for i in range(length):
    Y[i]=[ordered_values[i],1-ordered_values[i]]

  H=dict(zip(temp_ind,(Y)))
  
  return H
```

```python
def same_structure(ref, inp):
    # Case 1: types don't match
    if isinstance(ref, dict) != isinstance(inp, dict):
        return False

    # Case 2: both are dicts → compare keys
    if isinstance(ref, dict):
        if set(ref.keys()) != set(inp.keys()):
            return False
        
        # Recursively check nested dictionaries
        for key in ref:
            if not same_structure(ref[key], inp[key]):
                return False

        return True

    # Case 3: both are non-dicts → structural match
    return True
```

```python
def positive_potential(the_weight_dictionary, algebra_dict, t, logic_dictionary, color_dict, det_for_loop, s, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict,select_dict,temporary_zeroed_s_value_dict):
    ne=0
    while True:
        bk=0
        option1_array=list(the_weight_dictionary.keys())
        option1_input=input('The dictionary keys for weight is: \n'+str(option1_array)+'. What keys would you like to change? \nType the list for "keys" you would lie to modify. e.g.["'"temperature_preservation"'", "'"mobility"'",...]. The options are '+str(the_weight_dictionary.keys()))
        
        if option1_input=='back' or option1_input=='b':
            return the_weight_dictionary, logic_dictionary
        try:
            for option1 in ast.literal_eval(option1_input):
                if option1 in option1_array:
                    while True:
                        option2_input=input('For the chosen key '+str(option1)+', the dictionary is: '+str(the_weight_dictionary[option1])+'. Among the dictionary, which values would you like to change?\nType the list for keys you would lie to modify.')
                        option2_array=list(the_weight_dictionary[option1].keys())
                        if option2_input=='back' or option2_input=='b':
                            bk=1
                            break
                        try:
                            for option2 in ast.literal_eval(option2_input):
                                if option2 in option2_array:
                                    while True:  
                                        weight_input=input('How would you change the weight for '+str(option2)+'? Type the proper float. FYI, the original weight was ' +str(the_weight_dictionary[option1][option2]))
                                        if weight_input=='back' or weight_input=='b':
                                            bk=2
                                            break
                                        
                                        try:
                                            the_weight_dictionary[option1][option2]=float(weight_input)
                                    
                                            ne=1
                                            break
                                        except (ValueError, TypeError, SyntaxError):
                                            else_num=0                            
                                if bk==2:
                                    break    
                            if ne==1:
                                break


                        except (ValueError, TypeError, SyntaxError):
                            else_num=0
                        
                if bk==1:
                    break
                else:
                    else_num=0
       
        except (ValueError, TypeError, SyntaxError):
            else_num=0

        if ne==1:
            while True:
                generation_input=input("How long does this change last for? Type in any positive number or -1 for indefinite tow_s")
                if generation_input=='back' or generation_input=='b':
                    break
                try:
                    if int(generation_input)>0:
                        if 'positive_potential' in list(logic_dictionary['off'].keys()):
                            logic_dictionary['on']['positive_potential']=logic_dictionary['off']['positive_potential']
                            logic_dictionary['off'].pop('positive_potential')
                            logic_dictionary['on']['positive_potential']['generation']=int(generation_input)+t
                       
                        if 'positive_potential' in list(logic_dictionary['on'].keys()):
                            logic_dictionary['on']['positive_potential']['generation']=int(generation_input)+t
                            
                        return the_weight_dictionary, logic_dictionary
      
                    if int(generation_input)==-1:
                        if 'positive_potential' in list(logic_dictionary['off'].keys()):
                            logic_dictionary['on']['positive_potential']=logic_dictionary['off']['positive_potential']
                            logic_dictionary['off'].pop('positive_potential')
                            logic_dictionary['on']['positive_potential']['generation']=int(generation_input)+1000000
    
                        if 'positive_potential' in list(logic_dictionary['on'].keys()):
                            logic_dictionary['on']['positive_potential']['generation']=int(generation_input)+1000000
    
                        
                        return the_weight_dictionary, logic_dictionary
    
                    break
                    
                except(ValueError, TypeError):
                    else_num=0

        

def tow_potential(the_weight_dictionary, algebra_of_interest, t, logic_dictionary, s):

    tow_weight_dictionary=the_weight_dictionary['tow_s']
    
    if 'tow_s' in list(logic_dictionary['on'].keys()):

        matrix=algebra_of_interest
        depth, rows, cols = matrix.shape
               
        s_tow=0
        option=logic_dictionary['on']['tow_s']['option']
        
        if option=='fill_the_empty_y':
            
            for x in range(depth):
                for z in range(cols):
                    y_array=[]
                    for y in range(rows):
                        if matrix[x,y,z]==1:
                            y_array.append(y)

                    if y_array!=[]:
                        if len(y_array)!=max(y_array)-min(y_array):
                            s_tow+=len(y_array)*the_weight_dictionary['tow_s'][option]*s

            return s_tow

        
        if option=='higher_x':
            x_array=[]

            for y in range(rows):
                for z in range(cols):
                    for x in range(depth):
                        if matrix[x,y,z]==1:
                            x_array.append(x)
            count = sum(1 for x in x_array if x>=the_weight_dictionary['tow_s'][option]['criteria'])

            s_tow+=count*the_weight_dictionary['tow_s']['higher_x']['weight']*s
            return s_tow
        
        if option=='higher_z':
            z_array=[]
            for z in range(cols):
                for x in range(depth):
                    for y in range(rows):
                        if matrix[x,y,z]==1:
                            z_array.append(z)

            count=sum(1 for z in z_array if z>=the_weight_dictionary['tow_s'][option]['criteria'])

            s_tow+=count*the_weight_dictionary['tow_s']['higher_z']['weight']*s
            return s_tow
        
        if option=='higher_y':
            y_array=[]

            for y in range(rows):
                for z in range(cols):
                    for x in range(depth):
                        if matrix[x,y,z]==1:
                            y_array.append(y)

            count = sum(1 for y in y_array if y>=the_weight_dictionary['tow_s'][option]['criteria'])

            s_tow+=count*the_weight_dictionary['tow_s'][option]['weight']*s

            return s_tow
            
        if option=='selective_for_elements':
            element_array=[]
            for x in range(depth):
                for y in range(rows):
                    for z in range(cols):
                        for element in input_elements:
                            if matrix[x,y,z]==1:
                                element_array.append([x,y,z])

            s_tow+=len(element_array)*the_weight_dictionary['tow_s'][option]*s
                                
            return s_tow

    else:
        s_tow=0
        return s_tow
        
def tow_by_selection(the_weight_dictionary, algebra_dict, t, logic_dictionary, color_dict, det_for_loop, s, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict,select_dict,temporary_zeroed_s_value_dict):
    
    while True:
        possible_inputs=list(the_weight_dictionary['tow_s'].keys())+['f','fill','x','y','z','selective','s','elements','e']
        option=input("How would you like to tow_s your matrix? The options are: fill_the_empty_y, higher_x, higher_y, or higher_z")#, or selective for elements")
        if option=='back' or option=='b':
            return the_weight_dictionary, logic_dictionary

        if option in possible_inputs:
            if option=='fill_the_empty_y' or option=='f' or option=='fill':
                
                while True:
                    
                    weight_input=input("What is the weight for the tow_s for your option "+str(option)+"? FYI, the original weight was" +str(the_weight_dictionary['tow_s']['fill_the_empty_y']))
                    if weight_input=='back' or weight_input=='b':
                        option='0'
                        break
                    try: 
                        the_weight_dictionary['tow_s']['fill_the_empty_y']=float(weight_input)
                        print("yes1")
                        print("yes1 the_weight_dictionary", the_weight_dictionary)
                        while True:
                            generation_input=input("How long does this change last for? Type in any positive number or -1 for indefinite tow_s")
                            if generation_input=='back' or generation_input=='b':
                                break
                            try:
                                if int(generation_input)>0:
                                    if 'tow_s' in list(logic_dictionary['off'].keys()):
                                        logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                        logic_dictionary['off'].pop('tow_s')
                                        logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t
                                        logic_dictionary['on']['tow_s']['option']='fill_the_empty_y'
                                        print("yes2")
                                    if 'tow_s' in list(logic_dictionary['on'].keys()):
                                        logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t
                                    
                                if int(generation_input)==-1:
                                    if 'tow_s' in list(logic_dictionary['off'].keys()):
                                        logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                        logic_dictionary['off'].pop('tow_s')
                                        logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
                                        logic_dictionary['on']['tow_s']['option']='fill_the_empty_y'
           
                                    if 'tow_s' in list(logic_dictionary['on'].keys()):
                                        logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000

                                return the_weight_dictionary, logic_dictionary

                            except(ValueError, TypeError, SyntaxError):
                                else_num=0
                    
                            
                    except (ValueError, TypeError, SyntaxError):
                        else_num=0


            if option=='higher_x' or option=='x':
                while True:
                    
                    weight_input=input("What is the weight for the tow_s for your option higher x? FYI, the original weight was" +str(the_weight_dictionary['tow_s']['higher_x']))
                    if weight_input=='back' or weight_input=='b':
                        option='0'
                        break
                    try: 
                        the_weight_dictionary['tow_s']['higher_x']['weight']=float(weight_input)
                        while True:
                            the_input_x_criteria=input("What is the criteria for which x to regard high x?")
                            try:    
                                the_weight_dictionary['tow_s']['higher_x']['criteria']=int(the_input_x_criteria)
                                while True:
                                    generation_input=input("How long does this change last for? Type in any positive number or -1 for indefinite tow_s")
                                    if generation_input=='back' or generation_input=='b':
                                        break
                                    try:
                                        
                                        if int(generation_input)>0:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t
                                                logic_dictionary['on']['tow_s']['option']='higher_x'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t

                                            
                                        if int(generation_input)==-1:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
                                                logic_dictionary['on']['tow_s']['option']='higher_x'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
                                        
                                        return the_weight_dictionary, logic_dictionary
        
                                    except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                            except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                                    
                    except (ValueError, TypeError, SyntaxError):
                        else_num=0

                
            if option=='higher_y' or option=='y':

                while True:
                    
                    weight_input=input("What is the weight for the tow_s for your option higher y? FYI, the original weight was" +str(the_weight_dictionary['tow_s']['higher_y']))
                    if weight_input=='back' or weight_input=='b':
                        option='0'
                        break
                    try: 
                        the_weight_dictionary['tow_s']['higher_y']['weight']=float(weight_input)
                        while True:
                            the_input_y_criteria=input("What is the criteria for which y to regard high y?")
                            try:    
                                the_weight_dictionary['tow_s']['higher_y']['criteria']=int(the_input_y_criteria)
                        
                                while True:
                                    generation_input=input("How long does this change last for? Type in any positive number or -1 for indefinite tow_s")
                                    if generation_input=='back' or generation_input=='b':
                                        break
                                    try:
                                        if int(generation_input)>0:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t
                                                logic_dictionary['on']['tow_s']['option']='higher_y'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t

                                            
                                        if int(generation_input)==-1:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
                                                logic_dictionary['on']['tow_s']['option']='higher_y'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
        
                                        return the_weight_dictionary, logic_dictionary
        
                                    except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                            
                            except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                            
                    except (ValueError, TypeError, SyntaxError):
                        else_num=0
                
            if option=='higher_z' or option=='z':
                option='0'
                while True:
                    
                    weight_input=input("What is the weight for the tow_s for your option higher z? FYI, the original weight was" +str(the_weight_dictionary['tow_s']['higher_z']))
                    if weight_input=='back' or weight_input=='b':
                        break
                    try: 
                        the_weight_dictionary['tow_s']['higher_z']['weight']=float(weight_input)
                        while True:
                            the_input_z_criteria=input("What is the criteria for which z to regard high z")
                            try:    
                                the_weight_dictionary['tow_s']['higher_z']['criteria']=int(the_input_z_criteria)
                        
                                while True:
                                    generation_input=input("How long does this change last for? Type in any positive number or -1 for indefinite tow_s")
                                    if generation_input=='back' or generation_input=='b':
                                        break
                                    try:
                                        if int(generation_input)>0:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t
                                                logic_dictionary['on']['tow_s']['option']='higher_z'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+t

                                            
                                        if int(generation_input)==-1:
                                            if 'tow_s' in list(logic_dictionary['off'].keys()):
                                                logic_dictionary['on']['tow_s']=logic_dictionary['off']['tow_s']
                                                logic_dictionary['off'].pop('tow_s')
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
                                                logic_dictionary['on']['tow_s']['option']='higher_z'

                                            if 'tow_s' in list(logic_dictionary['on'].keys()):
                                                logic_dictionary['on']['tow_s']['generation']=int(generation_input)+1000000
        
                                        return the_weight_dictionary, logic_dictionary
    
                                    except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                            
                            except(ValueError, TypeError, SyntaxError):
                                        else_num=0
                            
                    except(ValueError, TypeError, SyntaxError):
                        else_num=0
        else:
            else_num=0

def emergence_potential(the_weight_dictionary, logic_dictionary, new_algebra_of_interest, s):
    s_emergence=0
    if 'emergence' in list(logic_dictionary['on'].keys()):
        a=the_weight_dictionary['emergence']['t**2']
        b=the_weight_dictionary['emergence']['m**2']
        c=the_weight_dictionary['emergence']['t*m']
        tp=temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)
        mo=mobility(the_weight_dictionary, new_algebra_of_interest, s)
        s_emergence=a*(tp**2)+b*(mo**2)+d*(tp*mo)
    return s_emergence
    
def emergence(the_weight_dictionary, logic_dictionary, t):
    ne=0
    while True:
        emergence_input=input('Would you like to activate the emergence? Type yes or no')
        if emergence_input=='back' or emergence_input=='b':
            return logic_dictionary, the_weight_dictionary
     
        if emergence_input=='yes' or emergence_input=='y':
            
            while True:                
                strength_input=input('What are the keys and values for the strength of emergence? Type in the dictionary with {key1: value1, key2: value2,...}. \nThe currently existing dictionary is '+str(the_weight_dictionary['emergence']))
                if strength_input=='back' or strength_input=='b':
                    break
                try:
                    cthe_weight_dictionary=copy.deepcopy(the_weight_dictionary['emergence'])
                    
                    the_weight_key_array=list(cthe_weight_dictionary.keys())
                    strength_dictionary=ast.literal_eval(strength_input)
                    strength_key_array=list(strength_dictionary.keys())
                    for key in strength_key_array:
                        if key in the_weight_key_array:
                            print("key",key)
                            
                            the_weight_dictionary['emergence'][key]=strength_dictionary[key]
                            ne=1
                        else:
                            break
                            
                    if ne==1:   
                        while True:
                            duration_input=input('How long should the emergence last? Type in any positive number or -1 for indefinite emergence')
                            if duration_input=='back' or duration_input=='b':
                                ne=0
                                break
                            try:
                                if 'emergence' in list(logic_dictionary['off'].keys()):
                                    if int(duration_input)>0:
                                        logic_dictionary['on']['emergence']=logic_dictionary['off']['emergence']
                                        logic_dictionary['off'].pop('emergence')
                                        logic_dictionary['on']['emergence']['generation']=int(duration_input)+t
                                    if int(duration_input)<0:
                                        logic_dictionary['on']['emergence']=logic_dictionary['off']['emergence']
                                        logic_dictionary['off'].pop('emergence')
                                        logic_dictionary['on']['emergence']['generation']=int(duration_input)+1000000
                                if 'emergence' in list(logic_dictionary['on'].keys()):
                                    if int(duration_input)>0:
                                        logic_dictionary['on']['emergence']['generation']=int(duration_input)+t
                                    if int(duration_input)<0:
                                        logic_dictionary['on']['emergence']['generation']=int(duration_input)+1000000
                                
                                return logic_dictionary, the_weight_dictionary
    
                            except (ValueError, TypeError, SyntaxError):
                                else_num=0

                except (ValueError, TypeError, SyntaxError):
                    else_num=0            
                
        if emergence_input=='no' or emergence_input=='n':

            return logic_dictionary, the_weight_dictionary

        else:
            else_num=0
            
    
def new_function_maker(surviving_ones):
    new_function={}
    weight=0
    for ones in list(surviving_ones.keys()):
        weight+=surviving_ones[ones]

    if weight!=0:
        for ones in list(surviving_ones.keys()):
            new_function[ones]=[surviving_ones[ones]/weight, 1-(surviving_ones[ones]/weight)] 
    
    return new_function


def all_indices_one(matrix, index_list):

    # Check that matrix is 3-dimensional
    if matrix.ndim != 3:
        raise ValueError("Matrix must be 3D")

    # Collect matrix shape
    x_max, y_max, z_max = matrix.shape

    for idx in index_list:
        # Convert to tuple for indexing
        idx_tuple = tuple(idx)

        # Check index validity
        if len(idx_tuple) != 3:
            raise ValueError(f"Invalid index {idx}: must have 3 elements for a 3D array.")

        x, y, z = idx_tuple

        # Bounds checking
        if not (0 <= x < x_max and 0 <= y < y_max and 0 <= z < z_max):
            raise IndexError(f"Index {idx} exceeds matrix bounds {matrix.shape}")

        # Check the value
        if matrix[x, y, z] != 1:
            return False  # Early exit → at least one nonzero value

    return True  # All designated indices are one
    
def all_indices_zero(matrix, index_list):
    """
    matrix: 3D numpy array
    index_list: list of index triplets, e.g. [[1,1,3], [1,2,5], [2,2,4]]
    
    Returns True if all designated positions are zero, False otherwise.
    """

    # Check that matrix is 3-dimensional
    if matrix.ndim != 3:
        raise ValueError("Matrix must be 3D")

    # Collect matrix shape
    x_max, y_max, z_max = matrix.shape

    for idx in index_list:
        # Convert to tuple for indexing
        idx_tuple = tuple(idx)

        # Check index validity
        if len(idx_tuple) != 3:
            raise ValueError(f"Invalid index {idx}: must have 3 elements for a 3D array.")

        x, y, z = idx_tuple

        # Bounds checking
        if not (0 <= x < x_max and 0 <= y < y_max and 0 <= z < z_max):
            raise IndexError(f"Index {idx} exceeds matrix bounds {matrix.shape}")

        # Check the value
        if matrix[x, y, z] != 0:
            return False  # Early exit → at least one nonzero value

    return True  # All designated indices are zero


def negative_potential(the_weight_dictionary, algebra_dict, t, logic_dictionary, color_dict, det_for_loop, s, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict,select_dict,temporary_zeroed_s_value_dict):
    while True:
        condition_input=input("How are you exerting your negative potential? If you want to exert it by choosing the necessarily required matrix element, type 'matrix required element'\n. If you want to exert it by choosing the should not exist matrix element, type 'matrix absence element'\n If you want to choose matrices that suffice some potentials, type 'potential'")
        
        
        if condition_input=='matrix required element' or condition_input=='required' or condition_input=='r':
            while True:
                element_input=input("Type the indices of the points in the list that should exist in order to survive. e.g., [[1,2,3],[1,2,4],...]")
                try:
                    element_list=ast.literal_eval(element_input)
                    break
                except (ValueError, TypeError, SyntaxError):
                    else_num=0

            
            check_array=[]
            surviving_ones={}
            for prime in list(function.keys()):
                matrix_of_interest=tuple_to_np_array(summed_algebra_dict[prime])
                if all_indices_one(matrix_of_interest, element_list)==True:
                    check_array.append(prime)
                        
            for prime in check_array:
                surviving_ones[prime]=function[prime][0]

                
            if surviving_ones!={}:
                new_function=new_function_maker(surviving_ones)
            if surviving_ones=={}:
                new_function={}            

            return new_function


        if condition_input=='matrix absence element' or condition_input=='absence' or condition_input=='a':
            while True:
                element_input=input("Type the indices of the points in the list that should not exist in order to survive. e.g., [[1,2,3],[1,2,4],...]")
                try:
                    element_list=ast.literal_eval(element_input)
                    break
                except (ValueError, TypeError, SyntaxError):
                    else_num=0
            check_array=[]
            surviving_ones={}
            for prime in list(function.keys()):
                matrix_of_interest=tuple_to_np_array(summed_algebra_dict[prime])
                if all_indices_zero(matrix_of_interest, element_list)==True:
                    check_array.append(prime)
                        
            for prime in check_array:
                surviving_ones[prime]=function[prime][0]

                
            if surviving_ones!={}:
                new_function=new_function_maker(surviving_ones)
            if surviving_ones=={}:
                new_function={}            

            return new_function

        
        
                                       
        if condition_input=='potential' or condition_input=='p':
            while True:
                potential_input=input("What is the potential that you would specifically like to deal with? The options are lift, thrust, mobility, temperature preservation, or total")
                
                if potential_input=='back' or potential_input=='b':
                    break
                
                if potential_input=='lift' or potential_input=='l':
                    s_lif_average=0
                    for prime in list(function.keys()):
                        s_lif_average+=the_weight_dictionary['lift']*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)
                    s_lif_average/=len(list(function.keys()))
                    while True:
                        try:
                            s_input=input("What is the critical s_value? FYI, the current average s_value is" +str(s_thr_average))
                            if s_input=='back' or s_input=='b':
                                bk=0
                                break
                            s_value=float(s_input)
                            break
                        except(ValueError, TypeError, SyntaxError):
                            else_num=0
                    surviving_ones={}

                    for prime in list(function.keys()):
                        if lift_force(the_weight_dictionary, tuple_to_np_array(summed_algebra_dict[prime]),s)*s*the_weight_dictionary['mobility']['weight_mobility_lift_in_mobility']*the_weight_dictionary['algebra_specifics_selection']['weight_mobility_in_algebra_specifics_selection']>=s_value:
                            surviving_ones[prime]=function[prime][0]
                    
                    if surviving_ones!={}:
                        new_function=new_function_maker(surviving_ones)
                    if surviving_ones=={}:
                        new_function={}            
                 
                    return new_function
                
                if potential_input=='thrust':
                    s_thr_average=0
                    for prime in list(function.keys()):
                        s_thr_average+=the_weight_dictionary['thrust']*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)
                    s_thr_average/=len(list(function.keys()))
                    while True:
                        try:
                            s_input=input("What is the critical s_value? FYI, the current average s_value is" +str(s_thr_average))
                            if s_input=='back' or s_input=='b':
                                bk=0
                                break
                            s_value=float(s_input)
                            break
                        except(ValueError, TypeError, SyntaxError):
                            else_num=0
                    surviving_ones={}
                    for prime in list(function.keys()):
                        if thrust_force(the_weight_dictionary, tuple_to_np_array(summed_algebra_dict[prime]), s)*s*the_weight_dictionary['mobility']['weight_mobility_thrust_in_mobility']*the_weight_dictionary['algebra_specifics_selection']['weight_mobility_in_algebra_specifics_selection']>=s_value:
                            surviving_ones[prime]=function[prime][0]
                            
                    if surviving_ones!={}:
                        new_function=new_function_maker(surviving_ones)
                    if surviving_ones=={}:
                        new_function={}            
                 
                    return new_function
                
                if potential_input=='mobility' or potential_input=='m':
                    s_mob_average=0
                    for prime in list(function.keys()):
                        s_mob_average+=the_weight_dictionary['mobility']*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)
                    s_mob_average/=len(list(function.keys()))
                    while True:
                        try:
                            s_input=input("What is the critical s_value? FYI, the current average s_value is" +str(s_mob_average))
                            if s_input=='back' or s_input=='b':
                                bk=0
                                break
                            s_value=float(s_input)
                            break
                        except (ValueError, TypeError, SyntaxError):
                            else_num=0
                    surviving_ones={}
                    for prime in list(function.keys()):
                        if the_weight_dictionary['mobility']['weight_mobility_in_algebra_specifics_selection']*mobility(the_weight_dictionary, new_algebra_of_interest, s)>=s_value:
                            surviving_ones[prime]=function[prime][0]
                            
                    if surviving_ones!={}:
                        new_function=new_function_maker(surviving_ones)
                    if surviving_ones=={}:
                        new_function={}            

                    return new_function
                
                
                if potential_input=='temperature preservation' or potential_input=='temperature' or potential_input=='temp':
                    s_temp_average=0
                    for prime in list(function.keys()):
                        s_temp_average+=the_weight_dictionary['temperature_preservation']*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)
                    s_temp_average/=len(list(function.keys()))
                    while True:    
                        try:
                            s_input=input("What is the critical s_value? FYI, the current average s_value is" +str(s_temp_average))
                            if s_input=='back' or s_input=='b':
                                bk=0
                                break
                            s_value=float(s_input)
                            break
                        except (ValueError, TypeError, SyntaxError):
                            else_num=0
                    surviving_ones={}
                    for prime in list(function.keys()):
                        if the_weight_dictionary['temperature_preservation']*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)>=s_value:
                            surviving_ones[prime]=function[prime][0]
                            
                    if surviving_ones!={}:
                        new_function=new_function_maker(surviving_ones)
                    if surviving_ones=={}:
                        new_function={}            

                    return new_function
                
                if potential_input=='total':
                    s_average=0
                    for prime in list(function.keys()):
                        s_average+=algebra_specifics_selection(the_weight_dictionary, logic_dictionary, t, summed_algebra_dict[prime], s)
                    s_average=s_average/len(list(function.keys()))
                    
                    while True:
                        try:
                            s_input=input("What is the critical s_value? FYI, the current average s_value is " +str(s_average))
                            if s_input=='back' or s_input=='b':
                                bk=0
                                break
                            s_value=float(s_input)
                            break
                        except (ValueError, TypeError):
                            else_num=0
                    surviving_ones={}
                    for prime in list(function.keys()):
                        if algebra_specifics_selection(the_weight_dictionary, logic_dictionary, t, summed_algebra_dict[prime], s)>=s_value:
                            surviving_ones[prime]=function[prime][0]
                            
                    if surviving_ones!={}:
                        new_function=new_function_maker(surviving_ones)
                    if surviving_ones=={}:
                        new_function={}            

                    return new_function
 
                else:
                    else_num=0
                    
        else:
            else_num=0    

def mutation_rate(N, m, logic_dictionary, t):
      
    while True:
        dec=0
        mutation_input=input("What is your new mutation rate? Current mutation rate is "+str(m)+" and N is "+str(N)+" so that N*m is "+str(N*m))
        if mutation_input=='back' or mutation_input=='b':
            return m, logic_dictionary
        
        try:
            new_m=float(mutation_input)
            while True:
                verify=input("Is "+str(new_m)+" really the mutation rate you want so that N*m is "+str(N*new_m)+" ? Type yes or no")
                if verify=='back' or verify.lower()=='b':
                    break
                if verify.lower()=='yes' or verify.lower()=='y':
                    while True:
                        duration_input=input('How long should the new mutation rate last? Type in any positive number or -1 for indefinitely lasting effect')
                        if duration_input=='back' or duration_input=='b':
                            break
                        try:
                            if 'mutation_rate' in list(logic_dictionary['off'].keys()):
                                if int(duration_input)>0:
                                    logic_dictionary['on']['mutation_rate']=logic_dictionary['off']['mutation_rate']
                                    logic_dictionary['off'].pop('mutation_rate')
                                    logic_dictionary['on']['mutation_rate']['generation']=int(duration_input)+t
                                    
                                if int(duration_input)<0:
                                    logic_dictionary['on']['mutation_rate']=logic_dictionary['off']['mutation_rate']
                                    logic_dictionary['off'].pop('mutation_rate')
                                    logic_dictionary['on']['mutation_rate']['generation']=int(duration_input)+1000000
                            if 'mutation_rate' in list(logic_dictionary['on'].keys()):
                                if int(duration_input)>0:
                                    logic_dictionary['on']['mutation_rate']['generation']=int(duration_input)+t
                                if int(duration_input)<0:
                                    logic_dictionary['on']['mutation_rate']['generation']=int(duration_input)+1000000    
                        
                            return new_m, logic_dictionary

                        except (ValueError, TypeError, SyntaxError):
                            else_num=0
                        
                if verify.lower()=='no' or verify.lower()=='n':
                    while True:
                        keep_go_or_not=input("Do you still want to change the mutation rate? Type yes or no")
                        if keep_go_or_not=='back' or keep_go_or_not.lower()=='b':
                            break
                        if keep_go_or_not.lower()=='yes' or keep_go_or_not.lower()=='y':
                            dec=1
                            break
                        if keep_go_or_not.lower()=='no' or keep_go_or_not.lower()=='n':
                            return m, logic_dictionary
                        else:
                            print("Type correct word")
                if dec==1:
                    print("Then choose the new mutation rate again")
                    break
                    
                else:
                    print("Type correct word")
                
        except (ValueError, TypeError, SyntaxError):
            print("Type correct float")   

def new_information_for_modification(the_weight_dictionary, algebra_dict, t, s, logic_dictionary, color_dict, select_dict, det_for_loop, new_matrix, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict):
    if prime_set_part==[]:
            
        if prime_set_array==[]:
            det_for_loop=-1
            return summed_algebra_dict, function, prime_index, prime_set_part, prime_set_array, det_for_loop
        
        else:
            prime_set_array.remove(prime_set_array[0])
            prime_set_part=copy.deepcopy(prime_sets[prime_set_array[0]])
            next_prime=prime_set_part[0]
            
    else:
        next_prime=prime_set_part[0]
        
    function.clear()
    function[next_prime]=[1,0]
    summed_algebra_dict[next_prime]=np_array_to_tuple(new_matrix)
    prime_set_part.remove(next_prime) 
    prime_set_part.sort()
    prime_index+=1
    algebra_dict.clear()
    algebra_dict[next_prime]=np_array_to_tuple(new_matrix)
    select_dict[next_prime]=algebra_specifics_selection(the_weight_dictionary, logic_dictionary, t, new_matrix, s)
    color_dict[next_prime]=color_yielding_function()

                                        
           
    return algebra_dict, color_dict, select_dict, summed_algebra_dict, function, prime_index, prime_set_part, prime_set_array, det_for_loop

def explicit_modification(the_weight_dictionary, algebra_dict, t, logic_dictionary, color_dict, det_for_loop, s, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict,select_dict,temporary_zeroed_s_value_dict):
    sorted_dict=sorted_dictionary_maker(function) #sorted_dict={prime, distn[0]} values are listed in descending order
    detectable_dict=detectable_dictionary(sorted_dict) #remaining only the dictionaries that suffice certain amount of population
    detectable_dict_keys=list(detectable_dict.keys()) 
    detectable_dict_number_array=[x for x in range(len(detectable_dict_keys))]
    for prime in detectable_dict_keys:
        print("\n\n\nnumber:",detectable_dict_keys.index(prime))
        matrix_to_3d(2, np.array(summed_algebra_dict[prime]))
        print("temperature_preservation",temperature_preservation(the_weight_dictionary, np.array(summed_algebra_dict[prime]), s))
        print("mobility",mobility(the_weight_dictionary, np.array(summed_algebra_dict[prime]), s))
        print("total_s:",select_dict[prime])
        print("temporary_zeroed_s_value:",temporary_zeroed_s_value_dict[prime])
        print("mass",np.sum(summed_algebra_dict[prime])*the_weight_dictionary['mobility']['mass_in_mobility'])

    while True:
        dec=0
        bk=0
        else_num=0
        choice1=input("Among the listed detectable algebras, what algebra do you want to choose and explicitly modify?")
        if choice1=='back' or choice1=='b':
            while True:
                choice7=input("Are you sure to get out of explicit modification? yes or no")
                if choice7=='yes' or choice7=='y':
                    return algebra_dict, color_dict, select_dict, summed_algebra_dict, function, prime_index, prime_set_part, prime_set_array, det_for_loop
                if choice7=='no' or choice7=='n':
                    break
                else:
                    else_num=0
        try:
            number=int(choice1)
            
            if number in detectable_dict_number_array:
                
                while True:
                    more=0
                    choice2=input("Is "+str(number)+" the right choice you intended? Type yes or no")
                    if choice2=='back' or choice2=='b':
                        bk=1
                        number=-1
                        break
                    if choice2.lower()=='yes' or choice2=='y':
                        matrix=np.array(summed_algebra_dict[detectable_dict_keys[number]])
                        matrix_to_3d(2, matrix)
                        depth, rows, cols = matrix.shape
                        
                        while True:
                            if more==1:
                                print(tuple_to_np_array(summed_algebra_dict[list(function.keys())[0]]).shape)
                                print(function.keys())
                                matrix=np.array(summed_algebra_dict[list(function.keys())[0]])
                                matrix_to_3d(2, matrix)
                                depth, rows, cols = matrix.shape
                            more=0
                            choice3=input("How do you want to change? Type the list of points that you want to convert it to 0 or 1. For example [[5,7,3],[5,7,4]]")
                            if choice3=='back' or choice3=='b':
                                bk=1
                                break
                           
                            try:
                                print("new_matrix_picture")
                                matrix_to_3d(2, matrix)
                                matrix1=copy.deepcopy(matrix)
                                matrix2=copy.deepcopy(matrix)
                                position_list_for_new_layer=ast.literal_eval(choice3)
                                for position in position_list_for_new_layer:
                                    if matrix1[tuple(position)]==0:
                                        matrix[tuple(position)]=1
                                    if matrix2[tuple(position)]==1:
                                        matrix[tuple(position)]=0
                                new_matrix=matrix
                                print("new_matrix_picture")
                                matrix_to_3d(2, new_matrix)
                                algebra_dict, color_dict, select_dict, summed_algebra_dict, function, prime_index, prime_set_part, prime_set_array, det_for_loop=new_information_for_modification(the_weight_dictionary, algebra_dict, t, s, logic_dictionary, color_dict, select_dict, det_for_loop, new_matrix, prime_index, prime_set_part, prime_set_array, prime_sets, function, summed_algebra_dict)
                                print("modification done")
                                del matrix1
                                del matrix2
                                
                                while True:
                                    choice6=input("Do you want to modify more? Type yes or no")
                                    if choice6=='back' or choice6=='b':
                                        bk=1
                                        break
                                    if choice6=='yes' or choice6=='y':
                                        more=1
                                        break
                                    if choice6=='no' or choice6=='n':
                                        return algebra_dict, color_dict, select_dict, summed_algebra_dict, function, prime_index, prime_set_part, prime_set_array, det_for_loop
                                    else:
                                        else_num=0
                                    
                            except (ValueError, TypeError, SyntaxError):
                                print("Type the right list")
                                else_num=0
                    else:
                        else_num=0

                if number==-1:
                    break
                
        except (ValueError, TypeError, SyntaxError):
            else_num=0
            print("Type correct integer number")
```

```python
#For drawing Muller Plot
#whole_history_summed={t:{algebra:distn}} 
def mat_for_plot(whole_history,generation):
    whole_index=whole_index_finder(whole_history,generation) # returns the collection of whole algebra throughout all generations
    Y=copy.deepcopy(whole_history) # copy of whole_history_summed
    original_history=copy.deepcopy(Y) #copy of whole_history_summed
    aligned_history={} 
    concatenated_aligned_history={}
    proportion_density_history={}
    new_2D_whole_history={}
    time_interval_dictionary={}
    new_line_whole_history=new_line_whole_history_making_function(whole_history, generation, whole_index)
    #new_line_whole_history returns {algebra: [portion array for each generation when the algebra existed]}

    for t in range(generation):
      aligned_history[t]=align_analysis(original_history, aligned_history, t)
#returns {generation: [existing algebra array]},, note that the algebra array has specific order to reflect the origin of mutation
    for t in range(generation):
      concatenated_aligned_history[t]=concatenate_function(original_history, aligned_history, t)
#returns {generation: {algebra: [distn]}}
    for t in range(generation):
      proportion_density_history[t]=density_giving_function(concatenated_aligned_history, t)
#returns {generation:{algebra: accumulated proportion for that algebra}}    
    print("pdh done")
    new_2D_whole_history=new_2D_whole_history_making_function(proportion_density_history, generation, whole_index)

    time_interval_dictionary = time_interval(generation, whole_history,whole_index)


    return new_line_whole_history, proportion_density_history, new_2D_whole_history, time_interval_dictionary


def concatenate_function(original_history, aligned_history, t): #returns {generation: {algebra: [distn]}}
  A=original_history[t]
  B=list(aligned_history[t]) #this would return "the keys" for the t generation's aligned history
  C={}

  for element in B:
    C[element]=A[element]

  return C #for some particular generation t, C={key in aligned_history[t]:distn for that key at the original_history[t]}

def density_giving_function(concatenated_aligned_history,t): #returns {generation:{algebra: accumulated proportion for that algebra}}    
  proportion_density={0:0}
  #concatenated_aligned_history[t]={key in aligned_history[t]:distn for that key at the original_history[t]}
  A=concatenated_aligned_history[t].keys()
  temp_set=(list(A)) #list for keys for concatenated_algined_history for t generation, 
  #which is composed of key values for aligned history [t]

  for i in list(concatenated_aligned_history[t]): #keys in aligned_history[t]
    temp_proportion_density={}
    temp_proportion_density[i]=concatenated_aligned_history[t][i][0] #the probability value
    if temp_set.index(i)==0:
      proportion_density[i]=temp_proportion_density[i]+proportion_density[0]
    else:
      n=temp_set[temp_set.index(i)-1]
      proportion_density[i]=temp_proportion_density[i]+proportion_density[n]

  return proportion_density

def whole_index_finder(whole_history,generation): # returns the collection of whole algebra in an ordered sequence

  A=set({})
  for t in range(generation):
    A=A.union(set(whole_history[t].keys())) #collection of all algebra throughout the generation

  whole_index=list(A)
  whole_index.sort() #ordering 

  return whole_index

def align_algorithm(temp_prior, prior, new_born): # returns the aligned array that reflects "the sequnce" of the origination of mutation

  element_dictionary={}

  for element in new_born:
    A=[]
    for i in prior:
      if element%i==0: #if the (new) element is a multiple of preexisting number i in prior array,
        A.append(i) #append that preexisting number 'i's to array A
    a=max(A) #a is a maximum value among that 'i's 
    element_dictionary[element]=a #element dictionary is composed with {element: maximum i that divides the element (i.e., a)}
    n=temp_prior.index(element_dictionary[element]) #n is an index for a
    m=temp_prior.index(element) #m is an index for element
    temp_prior.insert(n,element) #insert 'element' to the temp_prior array at the index n
    del temp_prior[m+1] #delete the array for m+1

  return temp_prior

def align_analysis(original_history, aligned_history, t): #returns {generation: [existing algebra array]},, note that the algebra array has specific order to reflect the origin of mutation

  if t==0:
    A=copy.deepcopy(original_history[0]) #original_history[0]-> generation=0 일 때의 initial {algebra:distn}
    temp_present=list(A.keys()) #only algebras
    temp_present.reverse() #chaning the order of the algebras
    present=temp_present 

    return present # if t=0, returns the algebras of the initial state with inverse order (to reflect how the algebras would be represented in the Muller plot)

  else: #when t>0
    A={}
    A=copy.deepcopy(original_history[t])
    current=list(A.keys()) 
      
    B={}
    B=copy.deepcopy(aligned_history[t-1])
    prior=list(B)

    new_born=set(current)-set(prior) #the algebra element that only t generation have compared to t-1
    diminished=set(prior)-set(current) #the algebra element that only t-1 generation have compared to t

    temp_prior=copy.deepcopy(prior) #copy of prior, type: list

    if new_born!=set({}): #if there exists a new algebra

      for new in list(new_born): 
        temp_prior.append(new) #add those new algebra to the set of prior algebras

      temp_present=align_algorithm(temp_prior, prior, new_born) #returns the aligned algebra list that reflects the origin of the mutation

      if diminished != set({}): #if there exists an diminished algebra
        for old in list(diminished):
          temp_present.remove(old) #remove from the array

        present=temp_present

      else:
        present=temp_present

      return present

    else: #when there is no new algebra
      temp_present=temp_prior
      if diminished != set({}):
        for old in list(diminished):
          temp_present.remove(old)

        present=temp_present

      else:
        present=temp_prior
      return present

def time_interval(generation, whole_history,whole_index): #returns {algebra: times t that that algebra existed}
  time_interval_dict={}

  for i in whole_index: #whole algebras
    time_interval=[]
    for t in range(generation):
      if i in (whole_history[t].keys()): #if the algebra i is in the t generation's dictionary's key values
        time_interval.append(t) #add that t to the time_interval
    time_interval_dict[i]=time_interval #{algebra: times t that that algebra existed}

  return time_interval_dict

def new_2D_whole_history_making_function(proportion_density_history,generation, whole_index):
  new_2D_whole_history={}
  print("2D_making_ftn started")
  for i in whole_index:
    new_part=[]
    for t in range(generation):
      if i in list(proportion_density_history[t].keys()):
        new_part.append(proportion_density_history[t][i])

    new_2D_whole_history[i]=new_part

  return new_2D_whole_history

def new_line_whole_history_making_function(whole_history, generation, whole_index):
#whole_history={t:{algebra:distn}}
  new_line_whole_history={}
  print("1D_making_ftn started")
  for i in whole_index:
    new_line_part=[]
    for t in range(generation):
      if i in list(whole_history[t].keys()):
        new_line_part.append(float(list((whole_history[t][i]))[0]))

    new_line_whole_history[i]=new_line_part

  return new_line_whole_history #{algebra: 


def drawing_one_dim(title, info_with_title, draw_directory_for_iteration_detail, new_line_whole_history, time_interval_dictionary, summed_color_dict, summed_algebra_dict, N, m, s, iter):
  plt.rcParams['font.family']='Arial'
  plt.figure(figsize = (15,7))
  plt.xticks(size=15)
  plt.yticks(size=15)
  plt.xlabel('generation',fontsize=25,labelpad=5)  
  plt.ylabel('proportion',fontsize=25,labelpad=20)
  print("actual_1D_started")
  
  for i in new_line_whole_history:
    clr=summed_color_dict[summed_algebra_dict[i]]

    a=max(new_line_whole_history[i])
    plt.plot(time_interval_dictionary[i],new_line_whole_history[i],color=clr,linestyle='-')

  plt.savefig(str(draw_directory_for_iteration_detail)+'no title one dim.png')
  plt.rcParams['font.family']='Arial'
  plt.title('one dim with_'+str(info_with_title))
  plt.savefig(str(draw_directory_for_iteration_detail)+'one dim without legend.png')  
  plt.legend(loc=10,fontsize="8")
  plt.savefig(str(draw_directory_for_iteration_detail)+'one dim.png')

  plt.close()
  
def drawing_two_dim(title, info_with_title, draw_directory_for_iteration_detail, new_2D_whole_history, time_interval_dictionary, generation, proportion_density_history, summed_color_dict, summed_algebra_dict, N, m, s, iter):
  index_set=list(new_2D_whole_history.keys())

  plt.rcParams['font.family']='Arial'
  plt.figure(figsize = (15,7))
  plt.xticks(size=15)
  plt.yticks(size=15)
  plt.xlabel('generation',fontsize=25,labelpad=5)
  plt.ylabel('proportion',fontsize=25,labelpad=20)
  print("actual_2D_started")


  time_range=[]

  for t in range(generation):
    time_range.append(t)


  for t in range(generation-1):
    temp_ind1=list(proportion_density_history[t].keys())
    temp_ind2=list(proportion_density_history[t+1].keys())

    for i in temp_ind1:
      if i==temp_ind1[-1]:
          break

      else:
        next_ind1=temp_ind1.index(i)+1
        next_key1=temp_ind1[next_ind1]

        if i in temp_ind2:
          if next_key1 in temp_ind2:
            next_ind2=temp_ind2.index(i)+1
            if next_ind2<len(temp_ind2)-1:
              next_key2=temp_ind2[next_ind2]
              plt.fill([time_range[t],time_range[t],time_range[t+1],time_range[t+1]],[proportion_density_history[t][i], proportion_density_history[t][next_key1],proportion_density_history[t+1][next_key2], proportion_density_history[t+1][i]],color=summed_color_dict[summed_algebra_dict[next_key1]])
            if next_ind2==len(temp_ind2)-1:
              plt.fill([t,t,t+1],[proportion_density_history[t][i],proportion_density_history[t][next_key1],proportion_density_history[t+1][temp_ind2[-2]]],color=summed_color_dict[summed_algebra_dict[next_key1]])
          else:
            plt.fill([t,t,t+1],[proportion_density_history[t][i],proportion_density_history[t][next_key1],proportion_density_history[t+1][i]],color=summed_color_dict[summed_algebra_dict[next_key1]])

        else:
          temp_num=0
          while i not in temp_ind2:
            next_ind3=temp_ind1.index(i)-1
            next_key3=temp_ind1[next_ind3]
            i=temp_ind1[next_ind3]
            temp_num=i
          plt.fill([t,t,t+1],[proportion_density_history[t][i],proportion_density_history[t][next_key1],proportion_density_history[t+1][temp_num]],color=summed_color_dict[summed_algebra_dict[next_key1]])  

  
  plt.rcParams['font.family']='Arial'
  plt.savefig(str(draw_directory_for_iteration_detail)+'no title two dim.png')
  plt.title(str(info_with_title))
  plt.savefig(str(draw_directory_for_iteration_detail)+'two dim.png')

  plt.close()
```

```python
def rgb_to_hex(r, g, b):
    if not all(0 <= val <= 255 for val in (r, g, b)):
        raise ValueError("RGB values must be between 0 and 255")
    return "#{:02X}{:02X}{:02X}".format(r, g, b)
```

```python
import re

def numpy_no_comma_to_python_list(text):
    """
    Convert a NumPy-style no-comma matrix into a valid Python list
    with commas between elements, rows, and layers.
    """
    lines = text.strip().splitlines()
    out = []

    for line in lines:
        stripped = line.strip()

        # If the line looks like a matrix row: [0 1 0 1 ...]
        if stripped.startswith("[") and stripped.endswith("]"):
            inside = stripped[1:-1].strip()

            # If inside is empty, keep as empty row
            if inside == "":
                out.append("[]")
            else:
                nums = inside.split()
                out.append("[" + ", ".join(nums) + "],")   # <- add comma for row separation
        else:
            out.append(stripped)

    # Join lines and fix bracket formatting
    result = "\n".join(out)

    # Insert commas between layers:
    # detect lines like "]]" followed by "[["
    result = re.sub(r'\]\]\s*\[\[', ']],\n[[', result)

    # Remove trailing comma after last row of a layer before ]] 
    result = re.sub(r',(\s*\])', r'\1', result)

    return result
```

```python
def point_selector(array):
    x_array = [el[1] for el in array]
    indices = []  # start index of each jigsaw triplet

    for i in range(len(x_array)-2):
        x1, x2, x3 = x_array[i:i+3]
        if x1 == x3 and abs(x1-x2) == 1:
            indices.append(i)
    return indices

def total_length(array):
    factor = math.sqrt(2/3)
    jigsaw_starts = point_selector(array)
    length = 0
    n = len(array)
    
    for i in range(n-1):
        p0, p1 = array[i], array[i+1]
        dist = math.sqrt((p1[0]-p0[0])**2 + (p1[1]-p0[1])**2)

        # scale segment if it's part of any triplet
        for s in jigsaw_starts:
            if i == s or i == s+1:
                dist *= factor
                break
        
        length += dist

    return length

    
def temperature_preservation(the_weight_dictionary, algebra_of_interest, s):
    y_criteria_for_temperature=the_weight_dictionary['temperature_preservation']['y_criteria_for_temperature']
    weight_for_temperature=the_weight_dictionary['temperature_preservation']['weight_for_temperature']
    colgebra_of_interest=copy.deepcopy(algebra_of_interest)
    depth, rows, cols = algebra_of_interest.shape
    calgebra_of_interest=copy.deepcopy(algebra_of_interest)
    temporary_record=[]
    for d in range(depth):
        for r in range(rows):
            for c in range(cols):
                if colgebra_of_interest[d][r][c]==1:
                    temporary_record.append([d,r,c])

        
    temperature_s=0
    for y in range(rows):
        temp_matrix=colgebra_of_interest[:,y,:]
        z_array=[]
        min_z=0
        max_z=0
        for z in range(cols):
            for x in range(depth):
                if temp_matrix[x,z]==1:
                    if ([x,y-1,z] in temporary_record) or ([x,y+1,z] in temporary_record):
                        if y_criteria_for_temperature-y>=0:
                            temperature_s+=(((y_criteria_for_temperature-y)**(1/4))*the_weight_dictionary['temperature_preservation']['y_direction']*s*weight_for_temperature)
                        else: 
                            temperature_s+=0
                    else:
                        if y_criteria_for_temperature-y>=0:
                            temperature_s+=(((y_criteria_for_temperature-y)**(1/4))*the_weight_dictionary['temperature_preservation']['else']*s*weight_for_temperature)
                        else: 
                            temperature_s+=0


    return temperature_s


def thrust_force(the_weight_dictionary, algebra_of_interest, s):
    weight_for_thrust=the_weight_dictionary['thrust_force']['weight_for_thrust_in_thrust_force'] #the pressure that wing could exert
    depth, rows, cols = algebra_of_interest.shape
    colgebra_of_interest=copy.deepcopy(algebra_of_interest)

    force_from_thrust=0
    for y in range(rows):
        temp_matrix=colgebra_of_interest[:,y,:]
        z_array=[]
        min_z=0
        max_z=0
        for z in range(cols):
            for x in range(depth):
                if temp_matrix[x,z]==1:
                    z_array.append(z)
        if z_array!=[]:
            max_z=max(z_array)
            min_z=min(z_array)

        if max_z-min_z>=4:
            force_from_thrust+=weight_for_thrust*x_projected_thrust(algebra_of_interest, s,the_weight_dictionary)
    
    return force_from_thrust

def x_projected_thrust(algebra_of_interest, s,the_weight_dictionary):
    depth, rows, cols = algebra_of_interest.shape
    calgebra_of_interest=copy.deepcopy(algebra_of_interest)
    area_set=set({})
    for x in range(depth):
        for y in range(rows):
            for z in range(cols):
                if calgebra_of_interest[x][y][z]==1:
                    area_set.add((y,z))

    z_set=set({})
    area_list=list(area_set)
    count=0
    for el in area_list:
        count+=(abs(el[1]-5)*the_weight_dictionary['thrust_force']['weight_for_z_characteristics']+1)
        z_set.add(el[1])

   
    for z in z_set:
        y_array=[]
        for el in area_list:
            if z==el[1]:
                y_array.append(el[0])

        count+=count_consecutive_groups(y_array)*the_weight_dictionary['thrust_force']['weight_for_y_characteristics']     

    return count

def count_consecutive_groups(arr):
    if not arr:
        return 0

    arr.sort()
    count = 0

    for i in range(1, len(arr)):
        if arr[i] == arr[i-1] + 1:
            count += 1

    return count
    
def check_inclinations(points):
    for i in range(len(points) - 1):
        x1, y1 = points[i]
        x2, y2 = points[i+1]

        dx = x2 - x1
        dy = y2 - y1

        if dx == 0:
            slope_abs = float('inf')
        else:
            slope_abs = abs(dy / dx)

        if slope_abs > 1:
            return 0

    return 1

    
    
def x_projected_area(algebra_of_interest, s):
    depth, rows, cols = algebra_of_interest.shape
    calgebra_of_interest=copy.deepcopy(algebra_of_interest)
    area_set=set({})
    for x in range(depth):
        for y in range(rows):
            for z in range(cols):
                if calgebra_of_interest[x][y][z]==1:
                    area_set.add((y,z))

    area_list=list(area_set)

    area=len(area_list)

    return area






def lift_force(the_weight_dictionary, algebra_of_interest, s):
    time_for_air_to_pass_through=1
    air_density=1.225
    weight_for_lift=air_density/(2*time_for_air_to_pass_through**2)
    depth, rows, cols = algebra_of_interest.shape
    colgebra_of_interest=copy.deepcopy(algebra_of_interest)

    
    lift_force=0
    for y in range(rows):
        pressure_difference=0
        temp_matrix=colgebra_of_interest[:,y,:]
        min_x_line_array=[]
        max_x_line_array=[]
        min_x_length=0
        max_x_length=0
        initial_lift_crit=0
        z_array=[]
        min_z=0
        max_z=0
        for z in range(cols):
            x_array=[]
            for x in range(depth):
                if temp_matrix[x,z]==1:
                    x_array.append(x)
                    z_array.append(z)
            if x_array!=[]:
                max_x=max(x_array)
                min_x=min(x_array)

            if x_array!=[]:
                min_x_line_array.append([z,min_x])
                max_x_line_array.append([z,max_x])
        if z_array!=[]:
            max_z=max(z_array)
            min_z=min(z_array)
        
        if min_x_line_array!=[] and max_x_line_array!=[]:    
            if abs(min_x_line_array[0][1]-max_x_line_array[0][1])>=2 and abs(min_x_line_array[-1][1]-max_x_line_array[-1][1])>=2:
                initial_lift_crit=1

        if len(min_x_line_array)>1:
            for ind in range(len(min_x_line_array)-1):
                min_x_length=total_length(min_x_line_array)
        if len(max_x_line_array)>1:
            for ind in range(len(max_x_line_array)-1):
                max_x_length=total_length(max_x_line_array)
                
        max_l=max(max_x_length, min_x_length)
        min_l=min(max_x_length, min_x_length)
        z_dif=max_z-min_z
        
        if max_x_length>=min_x_length>0 and check_inclinations(max_x_line_array)==1 and check_inclinations(min_x_line_array)==1 and initial_lift_crit==0:
            pressure_difference+=weight_for_lift*(((max_l/min_l)**2)-1)*(the_weight_dictionary['lift_force']['weight_for_velocity_in_lift_force']**2)

        if min_x_length>max_x_length>0 and check_inclinations(min_x_line_array)==1 and check_inclinations(max_x_line_array)==1 and initial_lift_crit==0:
            pressure_difference+=-1*weight_for_lift*(((max_l/min_l)**2)-1)*(the_weight_dictionary['lift_force']['weight_for_velocity_in_lift_force']**2)
            
        if max_z-min_z>=4:
            lift_force+=pressure_difference*((max_z)-(min_z))*1 #dA=((max_z)-(min_z))*1 
    
    
    return abs(lift_force)



def mobility(the_weight_dictionary, algebra_of_interest, s):     
    colgebra_of_interest=copy.deepcopy(algebra_of_interest)
    depth, rows, cols = colgebra_of_interest.shape
    
    temporary_record=[]
    the_x_points=[]

    for d in range(depth):
        for r in range(rows):
            for c in range(cols):
                if colgebra_of_interest[d][r][c]==1:
                    temporary_record.append([d,r,c])
                    the_x_points.append(d)
    
    if max(the_x_points)-min(the_x_points)<=5:
        mass=(np.sum(colgebra_of_interest)+600)*the_weight_dictionary['mobility']['mass_in_mobility']

    if max(the_x_points)-min(the_x_points)>5:
        mass=((np.sum(colgebra_of_interest)*5+600)*the_weight_dictionary['mobility']['mass_in_mobility'])
        
    g=the_weight_dictionary['mobility']['g_in_mobility']
    mass_force=mass*g
    weight_mobility_thrust=the_weight_dictionary['mobility']['weight_mobility_thrust_in_mobility']
    weight_mobility_lift=the_weight_dictionary['mobility']['weight_mobility_lift_in_mobility']
    weight_mobility_null=the_weight_dictionary['mobility']['weight_mobility_null_in_mobility']
    mobility_s=0
    tf=thrust_force(the_weight_dictionary, algebra_of_interest, s)
    lf=lift_force(the_weight_dictionary, algebra_of_interest, s)
    if tf<mass_force/2 and lf<=mass_force:
        mobility_s=weight_mobility_null*x_projected_area(algebra_of_interest, s)*s

    elif tf>=mass_force/2 and lf<=mass_force:
        mobility_s=weight_mobility_thrust*(tf-mass_force/2)*s
    
    elif tf>mass_force/2 and lf>mass_force:
        mobility_s=weight_mobility_lift*(lf-mass_force)*s+weight_mobility_thrust*(tf-mass_force/2)*s
  
    return mobility_s

def np_array_to_tuple(np_array_algebra):
    to_list_info=np_array_algebra.tolist()
    converted_matrix=tuple(tuple(tuple(x) for x in sublist) for sublist in to_list_info)

    return converted_matrix
    
def tuple_to_np_array(tuple_algebra):
    new_algebra_of_interest = []
    for inner in tuple_algebra:
        inner_list = []
        for x in inner:
            inner_list.append(list(x))
        new_algebra_of_interest.append(inner_list)

    new_algebra_of_interest=np.array(new_algebra_of_interest)

    return new_algebra_of_interest



def algebra_specifics_selection(the_weight_dictionary, logic_dictionary, t, algebra_of_interest, s):
    weight_temperature=the_weight_dictionary['algebra_specifics_selection']['weight_temperature_in_algebra_specifics_selection']
    weight_mobility=the_weight_dictionary['algebra_specifics_selection']['weight_mobility_in_algebra_specifics_selection']
    new_algebra_of_interest=algebra_of_interest
    if type(algebra_of_interest)!=np.ndarray:
        
        new_algebra_of_interest = []
        for inner in algebra_of_interest:
            inner_list = []
            for x in inner:
                inner_list.append(list(x))
            new_algebra_of_interest.append(inner_list)

        
        new_algebra_of_interest=np.array(new_algebra_of_interest)

    s_total=weight_temperature*temperature_preservation(the_weight_dictionary, new_algebra_of_interest, s)+weight_mobility*mobility(the_weight_dictionary, new_algebra_of_interest, s)+tow_potential(the_weight_dictionary, new_algebra_of_interest, t, logic_dictionary, s)+emergence_potential(the_weight_dictionary, logic_dictionary, new_algebra_of_interest, s)
    
    return s_total
#bookmark
def np_array_to_tuple(np_array_algebra):
    to_list_info=np_array_algebra.tolist()
    converted_matrix=tuple(tuple(tuple(x) for x in sublist) for sublist in to_list_info)

    return converted_matrix
    
def tuple_to_np_array(tuple_algebra):
    new_algebra_of_interest = []
    for inner in tuple_algebra:
        inner_list = []
        for x in inner:
            inner_list.append(list(x))
        new_algebra_of_interest.append(inner_list)

    new_algebra_of_interest=np.array(new_algebra_of_interest)

    return new_algebra_of_interest
```
